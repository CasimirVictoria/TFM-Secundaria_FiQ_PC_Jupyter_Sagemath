### 1. Definició de Variables i Integració Gaussiana
Calculant la integral definida $\int_{-\infty}^{\infty} e^{-z^2} dz = \sqrt{\pi}$:

```python
var('z t omega')
assume(omega > 0)
x = function('x')

f_gauss = exp(-z^2)
res_integral = integrate(f_gauss, z, -oo, oo)
print("1. Integral Gaussiana:", res_integral)
```

### 2. Resolució d'Equació Diferencial Ordinària (Oscil·lador Harmònic)
Resolució de $\frac{d^2 x(t)}{dt^2} + \omega^2 x(t) = 0$:

```python
eq_diferencial = diff(x(t), t, 2) + omega^2 * x(t) == 0
res_ode = desolve(eq_diferencial, x(t), ivar=t)
print("2. Solució EDO Oscil·lador Harmònic:", res_ode)
```

### 3. Valors i Vectors Propis de la Matriu de Pauli $\sigma_x$

```python
sigma_x = matrix([[0, 1], [1, 0]])
valors_propis = sigma_x.eigenvalues()
vectors_propis = sigma_x.eigenvectors_right()
print("3. Valors Propis Pauli X:", valors_propis)
print("4. Vectors Propis Pauli X:", vectors_propis)
```
